```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier 
from sklearn import tree 

music_data = pd.read_csv('music.csv')
music_data # to run the file
X = music_data.drop(columns=['genre']) 
Y = music_data['genre'] 
    
model = DecisionTreeClassifier ()
model.fit(X,Y) 

tree.export_graphviz(model, out_file ='music-recommender.dot',
                     feature_names = ['age', 'gender'], #so we can see the rules that are created based on age and gender
                     class_names = sorted(Y.unique()), #so it displays the class of each note without duplicates
                     label='all', #so all have readable labels
                     rounded = True, #rounded corners
                     filled= True)# so each box is filled with a color

#after i run this, i drag and drop the created file in vscode and add an extender (dot)
```

```python
import pandas as pd
music_data=pd.read_csv('music.csv')
music_data # to run the file
```

```python
#cleaning/ preparing the data , removing duplicates etc, for this project, we just have to seperate the input from the output (inpout: age,gender, output: genre )
X = music_data.drop(columns=['genre']) #it will create a new dataset without this column
Y = music_data['genre'] # a data set with only this column
Y
```

```python
from sklearn.tree import DecisionTreeClassifier #the most known machine learninh algorithm library

model = DecisionTreeClassifier ()
model.fit(X,Y) #input and output
predictions = model.predict([[21,1], [22,0]]) #the prediction i want it to make are the ones not on the data i gave it
predictions
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier #the most known machine learninh algorithm library
from sklearn.model_selection import train_test_split #make it split the data to training and testing
from sklearn.metrics import accuracy_score # to be able to calculate the accuracy of the predictions

music_data=pd.read_csv('music.csv')
music_data # to run the file
#cleaning/ preparing the data , removing duplicates etc, for this project, we just have to seperate the input from the output (inpout: age,gender, output: genre )
X = music_data.drop(columns=['genre']) #it will create a new dataset without this column
Y = music_data['genre'] # a data set with only this column
#2 variables for training , 2 for testing
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.2) # 20% of the data is used for testing


model = DecisionTreeClassifier ()
model.fit(X_train,Y_train) #input and output only for the training data
predictions = model.predict(X_test)
# predictions = model.predict([[21,1], [22,0]]) #the prediction i want it to make are the ones not on the data i gave it

score = accuracy_score(Y_test, predictions) #it checks the testing input data if the gave the same result as the testing output data
score #every time i press to run this, it takes random percents so the accuracy changes
```
